```python
def parse_date(value):
    for fmt in ("%Y-%m-%d", "%d/%m/%Y", "%d-%m-%Y"):
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.NaT


parsed_dates = sales_df["Date"].apply(parse_date)

print("Invalid Dates:")
display(sales_df[parsed_dates.isna()])

print("Non-standard Date Formats:")
display(
    sales_df[
        ~sales_df["Date"].str.fullmatch(r"\d{4}-\d{2}-\d{2}")
    ]
)
```

## Stage 6: Time-Series Sales Analysis

This stage investigates daily sales performance,
cumulative revenue, rolling averages and differences
between reporting periods.

```python
financial_df["Day"] = financial_df["Date"].dt.day
financial_df["Month"] = financial_df["Date"].dt.month
financial_df["Year"] = financial_df["Date"].dt.year

display(financial_df[
    ["Transaction_ID", "Date", "Day", "Month", "Year"]
])
```

```python
from pathlib import Path
import importlib
import src.analytics as analytics

file_path = Path(analytics.__file__)

code = file_path.read_text(encoding="utf-8")

# Remove the incorrect Stage 6 code
position = code.find("def analyse_daily_sales")

if position != -1:
    code = code[:position].rstrip()

# Add the corrected function
code += '''

def analyse_daily_sales(df):
    df = df.copy()
    df["Date"] = pd.to_datetime(df["Date"])

    daily_sales = df.groupby("Date").agg(
        Daily_Revenue=("Net_Revenue", "sum"),
        Daily_Profit=("Gross_Profit", "sum"),
        Transaction_Count=("Transaction_ID", "count")
    ).sort_index()

    calendar = pd.date_range(
        start="2026-01-01",
        end="2026-01-31",
        freq="D"
    )

    daily_sales = daily_sales.reindex(calendar)
    daily_sales.index.name = "Date"

    daily_sales["Transaction_Count"] = (
        daily_sales["Transaction_Count"].fillna(0).astype(int)
    )

    daily_sales["Cumulative_Revenue"] = (
        daily_sales["Daily_Revenue"].fillna(0).cumsum()
    )

    daily_sales["Rolling_3_Day_Average"] = (
        daily_sales["Daily_Revenue"]
        .rolling(window=3, min_periods=3).mean()
    )

    return daily_sales.reset_index()
'''

compile(code, str(file_path), "exec")
file_path.write_text(code, encoding="utf-8")

importlib.invalidate_caches()
importlib.reload(analytics)

print("File updated:", file_path)
print("Function available:", hasattr(analytics, "analyse_daily_sales"))
```
