## Stage 2: Data Quality Investigation

This stage identifies missing values, duplicate records,
invalid quantities, inconsistent product names, invalid
dates and other data quality issues before cleaning.

```python
missing_summary = pd.DataFrame({
    "Missing_Values": sales_df.isnull().sum(),
    "Missing_Percentage": sales_df.isnull().mean() * 100
})

display(missing_summary)
```

```python
print("Duplicate Transaction IDs:",
      sales_df["Transaction_ID"].duplicated().sum())

print("Completely Duplicated Records:",
      sales_df.duplicated().sum())
```

```python
print("Negative Quantities:")
display(sales_df[sales_df["Quantity"] < 0])

print("Zero Quantities:")
display(sales_df[sales_df["Quantity"] == 0])
```

```python
print("Missing Unit Prices:")
display(sales_df[sales_df["Unit_Price"].isna()])

print("Invalid Unit Prices:")
display(sales_df[sales_df["Unit_Price"] <= 0])

print("Invalid Unit Costs:")
display(sales_df[sales_df["Unit_Cost"] <= 0])
```

```python
print("Unique Products:")
print(sales_df["Product"].unique())

print("\nUnique Categories:")
print(sales_df["Category"].unique())

display(
    sales_df.groupby("Product")["Category"].unique()
)
```

```python
def parse_date(value):
    for fmt in ("%Y-%m-%d", "%d/%m/%Y", "%d-%m-%Y"):
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.NaT


parsed_dates = sales_df["Date"].apply(parse_date)

print("Invalid Dates:")
display(sales_df[parsed_dates.isna()])

print("Non-standard Date Formats:")
display(
    sales_df[
        ~sales_df["Date"].str.fullmatch(r"\d{4}-\d{2}-\d{2}")
    ]
)
```

```python
invalid_discounts = sales_df[
    ~sales_df["Discount_Percent"].between(0, 100)
]

display(invalid_discounts)

print("Unique Discounts:")
print(sales_df["Discount_Percent"].unique())
```

```python
product_categories = {
    "Laptop": "Computers",
    "Monitor": "Computers",
    "Smartphone": "Mobile",
    "Keyboard": "Accessories",
    "Mouse": "Accessories",
    "Printer": "Office Equipment"
}

expected_categories = (
    sales_df["Product"]
    .str.strip()
    .str.title()
    .map(product_categories)
)

category_errors = sales_df[
    sales_df["Category"] != expected_categories
]

display(category_errors)
```
